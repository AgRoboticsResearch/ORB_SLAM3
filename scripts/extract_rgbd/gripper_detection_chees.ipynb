```python
def generate_chees_board_img(inner_points_x, inner_points_y):
    img = np.zeros((inner_points_x+1, inner_points_y+1), np.uint8)
    for i in range(inner_points_x+1):
        for j in range(inner_points_y+1):
            if (i+j)%2 == 0:
                img[i, j] = 1
    return img
```

```python
img = generate_chees_board_img(3, 8)
plt.imshow(img, cmap='gray')
cv2.imwrite('./temp/chess_board38.png', img*255)
```
